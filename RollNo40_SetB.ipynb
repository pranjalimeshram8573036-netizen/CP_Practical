```python
import math

raw_kc = input("Enter the equilibrium constant Kc: ")
print("Raw input value:", raw_kc)
print("Raw input value data type:", type(raw_kc))

try:
    kc = float(raw_kc)
    print("Converted Kc value:", kc)
    print("Converted Kc value data type:", type(kc))

    if not math.isfinite(kc) or kc <= 0:
        print("Please enter a finite Kc value greater than zero.")
    else:
        # Coefficients of Kc*x^2 - (3*Kc + 1)*x + 2*Kc = 0
        a = kc
        b = -(3 * kc + 1)
        c = 2 * kc

        discriminant = b**2 - 4 * a * c

        print("\nQuadratic coefficients:")
        print(f"a = {a}")
        print(f"b = {b}")
        print(f"c = {c}")
        print(f"Discriminant = {discriminant}")

        root1 = (-b + math.sqrt(discriminant)) / (2 * a)
        root2 = (-b - math.sqrt(discriminant)) / (2 * a)

        print(f"Root 1 = {root1}")
        print(f"Root 2 = {root2}")

        meaningful_roots = [
            root for root in (root1, root2)
            if 0 <= root <= 1
        ]

        if not meaningful_roots:
            print("No physically meaningful root was found.")
        else:
            x = meaningful_roots[0]
            print(f"\nPhysically meaningful root: x = {x:.3f} mol/L")

            equilibrium_a = 2 - x
            equilibrium_b = 1 - x
            equilibrium_c = x

            print(f"Equilibrium [A] = {equilibrium_a:.3f} mol/L")
            print(f"Equilibrium [B] = {equilibrium_b:.3f} mol/L")
            print(f"Equilibrium [C] = {equilibrium_c:.3f} mol/L")

            conversion_percentage = (x / 1.0) * 100
            print(f"Conversion of limiting reactant B = {conversion_percentage:.2f}%")

            
            if conversion_percentage >= 70:
                print("Conversion is High.")
            elif conversion_percentage >= 40:
                print("Conversion is Moderate.")
            else:
                print("Conversion is Low.")

except ValueError:
    print("That isn't a valid number. Please enter a number for Kc.")
```

Queation 2 Part A)dont give code explain dc/dt=-Kc use the Euler method: C_new = C_old - Kc_old*Delta t cnod=1.5 mol/l k=0.05 min inverse, delta t=2 min using a for loop, simulate 20 mins and print a table of time, concentration (mol/l) and conversion. 2) using mat.exp(), calculate the exact value cnode to power (-kt) at 20 min and print the absolute error of the Euler result 3) using while loop repeat the simulation until conversion reachs 50% and time reaches 100 min print the time neded and the final concentration   

```python
import math

C0 = 1.5       # Initial concentration (mol/L)
k = 0.05        # Rate constant (min^-1)
delta_t = 2     # Time step (min)
total_time = 20 # Simulation duration (min)

concentration = C0
time = 0

print(f"{'Time (min)':>10} {'Concentration (mol/L)':>24} {'Conversion (%)':>18}")
print(f"{time:>10} {concentration:>24.3f} {0:>18.2f}")

for step in range(total_time // delta_t):
    concentration = concentration - k * concentration * delta_t
    time += delta_t
    conversion = (C0 - concentration) / C0 * 100
    print(f"{time:>10} {concentration:>24.3f} {conversion:>18.2f}")

exact_concentration = C0 * math.exp(-k * 20)


print(f"\nExact concentration at {total_time} min: {exact_concentration:.6f} mol/L")

# Error between the 20-minute Euler result and the exact result
absolute_error = abs(concentration - exact_concentration)
print(f"Absolute error: {absolute_error:.6f} mol/L")

# Start a fresh simulation for the while loop
concentration = C0
time = 0
conversion = 0

while conversion < 50 and time < 100:
    concentration = concentration - k * concentration * delta_t
    time += delta_t
    conversion = (C0 - concentration) / C0 * 100

print(f"Time needed: {time} min")
print(f"Final concentration: {concentration:.3f} mol/L")
print(f"Conversion: {conversion:.2f}%")
```

```python
tag = input("Enter a tag: ")
print("Tag:", tag)

if tag == "":
    print("Empty tag.")
else:
    digit_count = 0
    digit_sum = 0

    for character in tag:
        if character.isdigit():
            digit_count += 1
            digit_sum += int(character)

    if digit_count == 0:
        print("Tag with no digits.")
    else:
        print("Number of digits:", digit_count)
        print("Sum of digits:", digit_sum)
```

Question 3) dont give code a laboratory has n candidates catalysts and wants to test r of them togrther order does not matter write a recursive function of factorial n base case 0 factorial  function combination is n,r = nfactorial/n(r!(n-r)!) reject invalid input

```python
def factorial(number):
    if number == 0:  # Base case: 0! = 1
        return 1
    return number * factorial(number - 1)


def combination(n, r):
    return factorial(n) // (factorial(r) * factorial(n - r))


try:
    n = int(input("Enter the number of candidates (n): "))
    r = int(input("Enter how many to test together (r): "))

    if n < 0 or r < 0 or r > n:
        print("Invalid input: enter n >= 0 and 0 <= r <= n.")
    else:
        print(f"Number of combinations: {combination(n, r)}")

except ValueError:
    print("Invalid input: please enter whole numbers.")
```
